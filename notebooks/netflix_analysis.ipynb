Netflix data analysis

```python
import pandas as pd
from sqlalchemy import create_engine
```

```python
# Database connection variables
username = "root"
password = ""
host = "localhost"   # or remote host IP
port = "3306"        # default MySQL port
database = "data_analysis"
```

```python
conn = create_engine(f"mysql+pymysql://{username}:{password}@{host}:{port}/{database}")
```

```python
netflix_data = pd.read_sql("select * from netflix;", conn)
```

```python
netflix_data.loc[netflix_data['release_year'] > 2015].sort_values(by='release_year').head(10)
```

```python
pd.read_sql("SELECT * from netflix WHERE release_year > 2015 ORDER By release_year ASC LIMIT 10;", conn)
```
