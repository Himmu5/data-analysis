#### Retail Sale Data

```python
import pandas as pd
from sqlalchemy import create_engine
```

```python
# Database connection variables
username = "root"
password = ""
host = "localhost"   # or remote host IP
port = "3306"        # default MySQL port
database = "data_analysis"
```

```python
conn = create_engine(f"mysql+pymysql://{username}:{password}@{host}:{port}/{database}")
```

```python
retail_sale = pd.read_sql("select * from retail_sale;", conn)
```

```python
pd.read_sql("select * from retail_sale LIMIT 10;", conn)
```

#### Retrieve all columns for sales made on 2022-11-05.

```python
pd.read_sql("select * from retail_sale WHERE sale_date = '2022-11-05 LIMIT 100;';", conn)
```

#### Retrieve all transactions where the category is 'Clothing' and the quantity sold is more than 4 in Nov-2022.

```python
pd.read_sql("SELECT * FROM retail_sale WHERE category = 'Clothing' AND quantity > 4 AND MONTH(sale_date) = 11 AND YEAR(sale_date) = 2022;", conn)
```

Calculate the total sales (total_sale) for each category.

```python
pd.read_sql("SELECT category, SUM(total_sale) as Total_Sale FROM retail_sale GROUP BY category;", conn)
```

```python
retail_sale.groupby("category")['total_sale'].sum().reset_index(name='Total_Sale')
```

Find the average age of customers who purchased items from the 'Beauty' category.

```python
result = pd.read_sql("SELECT AVG(age) AS AVG_AGE FROM retail_sale WHERE category = 'Beauty';", conn)
result
```

```python
retail_sale.loc[retail_sale['category'] == 'Beauty'].aggregate("age").mean()
```

Retrieve all transactions where the total_sale is greater than 1000.

```python
result = pd.read_sql("SELECT * FROM retail_sale WHERE total_sale > 1000;", conn);
result
```

```python
retail_sale.loc[retail_sale['total_sale'] > 1000]
```

Find the total number of transactions made by each gender in each category.

```python
pd.read_sql("SELECT gender, COUNT(transactions_id) FROM retail_sale GROUP BY gender;", conn)
```

```python
retail_sale.groupby("gender").aggregate("transactions_id").count().reset_index()
```
