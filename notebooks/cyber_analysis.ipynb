### Find the top 5 users with the most login attempts.

```python
auth_log_data['username'].value_counts().reset_index().head()
```

### Count the number of failed vs successful logins per user.

```python
auth_log_data.aggregate("status").value_counts().reset_index()
```

### Identify the IP addresses with more than 50 failed login attempts in a day.

```python
auth_log_data.aggregate('source_ip').value_counts().reset_index()
```

### Find the first and last login time of each user.

```python
auth_log_data['timestamp'] = pd.to_datetime(auth_log_data['timestamp'])
sorted_auth_data = auth_log_data.sort_values(by=['log_id','timestamp'])
sorted_auth_data.groupby("username").first()
```

```python
sorted_auth_data.groupby("username").last()
```
