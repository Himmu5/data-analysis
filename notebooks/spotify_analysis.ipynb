### List all unique artists in the dataset.

```python
pd.read_sql("select COUNT(DISTINCT Artist) as count from spotify ;", con=conn)
```

### Find the total number of tracks available.

```python
pd.read_sql("SELECT COUNT(DISTINCT Track) as total_tracks From spotify;", con=conn)
```

### Show the top 10 tracks with the highest number of streams.

```python
pd.read_sql("SELECT * From spotify ORDER BY Stream DESC LIMIT 10;", con=conn);
```

### Count how many tracks belong to each album type (e.g., album, single).

```python
pd.read_sql("select Album_type, COUNT(Track) as count from spotify GROUP BY Album_type", con=conn)
```

### Retrieve all tracks where the energy score is greater than 0.8.

```python
pd.read_sql("select Track, Energy from spotify WHERE Energy > 0.8 ORDER BY Energy DESC;", con=conn)
```

### Find the average danceability score for each artist.

```python
pd.read_sql("select Artist, AVG(Danceability) as avg_danceability from spotify GROUP BY Artist;", con=conn)
```

### Get the top 5 albums with the highest total views on YouTube.

```python
pd.read_sql("select Album, SUM(Views) as total_views from spotify WHERE most_playedon = 'Youtube' GROUP BY Album Order by total_views desc LIMIT 5;", con=conn)
```

### Find the track with the longest duration for each artist.

```python
sort_on_duration = pd.read_sql("select Artist, Max(Duration_min) as duration from spotify group by Artist order by duration desc LIMIT 1;", con=conn);
sort_on_duration
```
