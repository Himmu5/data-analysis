### List all unique artists in the dataset.

```python
pd.read_sql("select COUNT(DISTINCT Artist) as count from spotify ;", con=conn)
```

### Find the total number of tracks available.

```python
pd.read_sql("SELECT COUNT(DISTINCT Track) as total_tracks From spotify;", con=conn)
```

### Show the top 10 tracks with the highest number of streams.

```python
pd.read_sql("SELECT * From spotify ORDER BY Stream DESC LIMIT 10;", con=conn);
```

### Count how many tracks belong to each album type (e.g., album, single).

```python
pd.read_sql("select Album_type, COUNT(Track) as count from spotify GROUP BY Album_type", con=conn)
```

### Retrieve all tracks where the energy score is greater than 0.8.

```python
pd.read_sql("select Track, Energy from spotify WHERE Energy > 0.8 ORDER BY Energy DESC;", con=conn)
```

### Find the average danceability score for each artist.

```python
pd.read_sql("select Artist, AVG(Danceability) as avg_danceability from spotify GROUP BY Artist;", con=conn)
```

### Get the top 5 albums with the highest total views on YouTube.

```python
pd.read_sql("select Album, SUM(Views) as total_views from spotify WHERE most_playedon = 'Youtube' GROUP BY Album Order by total_views desc LIMIT 5;", con=conn)
```

### Find the track with the longest duration for each artist.

```python
sort_on_duration = pd.read_sql("select Artist, Max(Duration_min) as duration from spotify group by Artist order by duration desc LIMIT 1;", con=conn);
sort_on_duration
```

### Compare the average likes-to-views ratio between Spotify and YouTube (most_playedon).

```python
pd.read_sql("SELECT most_playedon, AVG(Views) as avg_views, AVG(Likes) as avg_likes, AVG(Views) * 1.0 / AVG(Likes) as ration from spotify group by most_playedon;", con=conn)
```

### Find the top 10 most commented tracks and their respective channels.

```python
commented = pd.read_sql("SELECT Album, Track, sum(Comments) as total_comment from spotify GROUP BY Track, Album ORDER BY total_comment DESC LIMIT 10;", con=conn);
commented
```

### For each artist, find the track with the maximum "energy × liveness" score.

```python
max_energy = pd.read_sql("select Track, MAX(energy * liveness) from spotify GROUP BY Track;", con=conn);
max_energy
```

### Identify whether tracks labeled as official video = True perform better in terms of average likes compared to official video = False.

```python
pd.read_sql("select official_video, SUM(Views) as total_views, SUM(Likes) as total_likes from spotify Group by official_video;", con=conn)
```

### Find the correlation between danceability and valence for all tracks (hint: use a correlation calculation or approximate with grouping).

```python
correlation_data = pd.read_sql("SELECT (AVG(Danceability * Valence) - AVG(Danceability) * AVG(Valence)) / (sqrt(AVG(Danceability * Danceability) - AVG(Danceability) * AVG(Danceability)) * sqrt(AVG(Valence * Valence) - AVG(Valence) * AVG(Valence))) AS correlation FROM spotify;",con=conn);
correlation_data
```

### Determine the top 5 artists who have the most consistent loudness (least variance across their tracks).

```python
pd.read_sql("SELECT Artist, COUNT(*) AS track_count, VARIANCE(Loudness) AS loudness_variance FROM spotify GROUP BY Artist HAVING COUNT(*) > 1 ORDER BY loudness_variance ASC LIMIT 5;", con=conn)
```
