### Linkedin Data Analysis

```python
import pandas as pd
from sqlalchemy import create_engine
from dotenv import load_dotenv
import os
```

```python
load_dotenv()
```

```python
jobs_data = pd.read_csv("../dataset/linkedin_jobs.csv")
jobs_data
```

```python
database = os.getenv("database")
port = os.getenv("port")
host = os.getenv("host")
password = os.getenv("password")
username = os.getenv("username")
```

```python
conn = create_engine(f"mysql+pymysql://{username}:{password}@{host}:{port}/{database}")
```

```python
# jobs_data.to_sql(con=conn, index=True, name="linkedin_jobs", if_exists="replace")
```

```python
pd.read_sql("select * from linkedin_jobs;", con=conn)
```

```python
pd.read_sql("select title, Count(title) from linkedin_jobs group by title;", con=conn)
```

### Find the companies hiring for "Software Engineer" roles specifically.

```python
pd.read_sql("select * from linkedin_jobs where title = 'Software Engineer';", con=conn)
```

```python
pd.read_sql("select * from linkedin_jobs where title Like '%% developer %%' LIMIT 10;", con=conn)
```

### Find the top 5 companies hiring in Bengaluru.

```python
pd.read_sql("select * from linkedin_jobs WHERE location LIKE '%%Bengaluru %%';", con=conn)
```
